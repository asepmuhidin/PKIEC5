# Tipe Data Koleksi



## List

### Create new List

```python
# Membuat List : nama_list[item1,item2..]
students=['Akmal Dimas','Alief Permana','Anindra','Dewi']
```

### Akses List

```python
# Untuk mengakses sebuah item list : nama_list[index]
print(f"Haloo nama saya {students[2]}")
```

### Multiple data type

```python
bio_student=['Akmal',23,170,'akmal@gmail.com',['Berenang','Main games']]
#Menampilkan hoby akmal
print(bio_student[4])
```

```python
#Latihan-1: Tampilkan output seperti ini : Nama akmal hobi berenang dan main games
print(f"Nama {bio_student[0]} hobi {bio_student[4][0]} dan {bio_student[4][1]} ")
```

### Item List Updated

```python
# Cara merubah nilai item list adalah : nama_list[index]=new_value
# Contoh sebelum update
print(f"Umur {bio_student[0]} adalah {bio_student[1]} ")

#Update umur akmal jadi 21
bio_student[1]=21

# Setelah update
print(f"Umur {bio_student[0]} adalah {bio_student[1]} ")
```

### Menampilkan Semua data List

```python
for student in students:
  print(student)
```

```python
# Menampilkan data list dan index list
for nomor,student in enumerate(students):
  print(nomor,student)
```

```python
# Latihan 2: Bagaimana code nya untuk output:
#1.Akmal Dimas
#2.Alief Permana
#3.Anindra
#4.Dewi
for nomor,student in enumerate(students):
  print(f"{nomor+1}.{student}")
```

### List operations

```python
# Operasi penjumlahan/ penggabungan
a=[1,2,3]
b=[4,5,6]

c=a+b

print(c)

#Penjumlahan pada LIST bukan operasi matrik
```

```python
# Operasi perkalian / perulangan
a=['Farhan','Imam']
daftar=a*3

print(daftar)
```

```python
# Pada dasarnya list seperti string
print("Farhann..."*10)
```

### List slices (Irisan list)

```python
# Menampikan seluruh data
print(students[:])
```

```python
# Menampilkan dari data pertama sampai data ke-n
# nama_list[:n]--> yg tampil data pertama sampai data ke n-1
print(students[:3])
```

```python
# Menampilkan dari data-m sampai data ke-n
# nama_list[m:n]--> yg tampil data ke-m sampai data ke n-1

print(students[1:3])
```

```python
# Menampilkan data terakhir: nama_list[-1]
print(students[-1])
```

### List methods

```python
# Menambahkan item list : nama_list.append(item)
students.append("Eza")
print(students[:])
```
