```python
bil=float(input("masukkan bilangan  ?"))
if bil>0:
  hasil="positif"
elif bil<0:
  hasil="negatif"
else:
  hasil="0"
print(f"{bil} adalah bilangan {hasil}")
```

```python
import math
a=float(input("a ?"))
b=float(input("b ?"))
c=float(input("c ?"))

d=b**2-4*a*c

if d>=0:
  x1=(-b+math.sqrt(d))/(2*a)
  x2=(-b-math.sqrt(d))/(2*a)
else:
  x1="x1 adalah akar imajiner"
  x2="x2 adalah akar imajiner"

print(f"x1={x1}")
print(f"x2={x2}")
print(f"d={d}")
```
