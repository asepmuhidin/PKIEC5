```python
def uang_lembur(gapok, jam_lembur):
  x=(gapok/173)*(1.5+((jam_lembur-1)*2))
  print (f"uang lembur anda adalah {x:,.2f}")

uang_lembur(5600000,28)
uang_lembur(6500000,28)
```

```python
def aktual_lembur(jam_lembur):
  return 1.5+(jam_lembur-1)*2

def uang_lembur(gapok, jam_lembur):
  x=(gapok/173)*(aktual_lembur(jam_lembur))
  return x

gaji_pokok=int(input("Masukkan gaji pokok anda ?"))
jam_lembur=int(input("Masukkan Jam lembur anda ?"))
print (f"uang lembur anda adalah {uang_lembur(gaji_pokok,jam_lembur):,.2f}")
```
