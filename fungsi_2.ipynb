```python
import math
def hitung_luas_lingkaran(r):
  return math.pi*r**2
```

```python
jari2_1=5
jari2_2=10

luas1=hitung_luas_lingkaran(jari2_1)
luas2=hitung_luas_lingkaran(jari2_2)

print(f"Luas lingkaran dengan jari-jari {jari2_1} adalah {luas1:,.2f} ")
print(f"Luas lingkaran dengan jari-jari {jari2_2} adalah {luas2:,.2f} ")
print(f"Jumlah Luas lingkaran adalah {luas1+luas2:,.2f} ")
```

```python
#Ini akan error, karena fungsi tersebut memerlukan nilai parameter 'r'
hitung_luas_lingkaran()
```

```python
def buat_profile(nama,umur,kota="jakarta", pekerjaan="belum bekerja"):
  print(f"====={nama.upper()}=====")
  print(f"Umur      : {umur}")
  print(f"Kota      : {kota}")
  print(f"Pekerjaan : {pekerjaan}")
  print()
```

```python
buat_profile("Farel", 20)
buat_profile("Farhan Fadilah", 21,"Bandung","Wiraswasta")
buat_profile(umur=22,pekerjaan="Mahasiswi", kota="Medan", nama="Dewi sartika")
```

```python
x=20
def test():
  x=10
  print(f"Nilai variabel x lokal adalah {x}")

test()
print(f"Nilai variabel x global adalah {x}")
```

```python
x=20
def test():
  global x
  x=10
  print(f"Nilai variabel x lokal adalah {x}")

test()
print(f"Nilai variabel x global adalah {x}")
```

```python
def jumlah(*bil):
  sum=0
  for i in bil:
    sum +=i

  print(f"Jumlah dari {bil} adalah {sum}")

jumlah(12,1,3,25,34)
```

```python
jumlah(2,3,5)
```
