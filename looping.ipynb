```python
for i in range(5):
  print(i, end=",")
```

```python
for i in [0,1,2,3,4]:
  print(i,end=",")
```

```python
for i in [10,11,12,13,14]:
  print(i,end=",")
```

```python
#Fungsi range(start,end,step)
for i in range(10,15):
  print(i, end=",")
```

```python
#Deret bilangan ganjil : 1,3,5,...15
for i in range(1,16,2):
  print(i,end=",")
```

```python
#Coba berlatih deret bilangan genap dari 8,10,12,...101
for i in range(8,101,2):
  print(i, end=",")
```

```python
#Deret 1,4,9,16,25,36,49...100
for i in range(1,11):
  print(i**2, end=",")
```

```python
#Menjumlah nilai dalam perulangan : 0+1+2+3+4
hasil=0
for i in range(5):
  hasil +=i    #hasil=hasil+i

print(f"Jumlah nilai dalam loop adalah {hasil}")
```

```python
nilai=[90,80,75,80,70]
hasil=0
for i in nilai:
  hasil +=i    #hasil=hasil+i

print(f"Jumlah nilai dalam loop adalah {hasil}")
```

```python
#Perulangan menurun, contoh 5,4,3,2,1
for i in range(5,0,-1):
  print(i,end=",")
```

```python
#Perkalian dalam perulangan, contoh bilangan faktorial
#3!=3x2x1
```

```python
hasil=1
n_fak=8
for i in range(n_fak,0,-1):
  hasil *=i
print(f"Nilai {n_fak}! adalah {hasil}")
```

```python
x=1
while x<=5:
  print(x, end=",")
  x+=1
```

```python
#Dengan while buat deret 10,12,14,..20
x=10
while x<=20:
  print(x,end=",")
  x+=2
```

```python
coba_lagi="y"
while coba_lagi=="y":
  hasil=1
  n_fak=int(input("Masukkan n! faktorial ?"))
  for i in range(n_fak,0,-1):
    hasil *=i

  print(f"Nilai {n_fak}! adalah {hasil}")
  coba_lagi=input("Mau mencoba lagi (y/t)?")
```
